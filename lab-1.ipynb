# Задание 2

```python
import cv2
from google.colab.patches import cv2_imshow


image = cv2.imread('/content/lando-norris.jpg')

cv2_imshow(image)
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

edges = cv2.Canny(gray, threshold1=100, threshold2=300)

contours, _ = cv2.findContours(edges, cv2.RETR_LIST, cv2.CHAIN_APPROX_SIMPLE)

output = image.copy()

cv2.drawContours(output, contours, -1, (0, 0, 255), 1)

cv2_imshow(output)
```

# Задание 3

```python
def draw_rectangle(img, top_left, bottom_right, color, thickness=1):
    max_y, max_x = img.shape[:2]
    bottom_right = (min(bottom_right[0], max_x-1), min(bottom_right[1], max_y-1))
    top_left = (max(top_left[0], 0), max(top_left[1], 0))

    for i in range(thickness):
        if top_left[1] + i < max_y and bottom_right[1] + i < max_y:
            img[top_left[1] + i:bottom_right[1] + i, top_left[0]] = color
            img[top_left[1] + i:bottom_right[1] + i, bottom_right[0]] = color

    for i in range(thickness):
        if top_left[0] + i < max_x and bottom_right[0] + i < max_x:
            img[top_left[1], top_left[0] + i:bottom_right[0] + i] = color
            img[bottom_right[1], top_left[0] + i:bottom_right[0] + i] = color


top_left = (700, 590)
bottom_right = (1300, 900)
color = (0, 0, 255)

draw_rectangle(image, top_left, bottom_right, color, 20)

cv2_imshow(image)
```
